```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load the Iris dataset
iris_data = load_iris(as_frame=True)
iris = iris_data.frame

# Add species names
iris['species'] = iris['target'].map(
    dict(enumerate(iris_data.target_names))
)

# Remove the numerical target column
iris = iris.drop(columns=['target'])

print("First five rows:")
print(iris.head())

print("\nDataset shape:", iris.shape)
print("\nSpecies count:")
print(iris['species'].value_counts())

# 1. Histogram
plt.figure(figsize=(7, 5))
sns.histplot(data=iris, x='sepal length (cm)', bins=15)
plt.title("Histogram of Sepal Length")
plt.show()

# 2. Scatter Plot
plt.figure(figsize=(7, 5))
sns.scatterplot(
    data=iris,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species'
)
plt.title("Sepal Length vs Sepal Width")
plt.show()

# 3. Box Plot
plt.figure(figsize=(7, 5))
sns.boxplot(
    data=iris,
    x='species',
    y='petal length (cm)'
)
plt.title("Petal Length by Species")
plt.show()

# 4. Pair Plot
sns.pairplot(iris, hue='species')
plt.suptitle("Pair Plot of Iris Dataset", y=1.02)
plt.show()

# 5. Bar Plot
plt.figure(figsize=(7, 5))
sns.barplot(
    data=iris,
    x='species',
    y='petal length (cm)',
    errorbar=None
)
plt.title("Average Petal Length by Species")
plt.show()

# 6. Correlation Heatmap
plt.figure(figsize=(7, 5))
correlation = iris.select_dtypes(
    include='number'
).corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```
