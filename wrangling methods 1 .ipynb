```python

import pandas as pd

# Create DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ravi', 'Sita', 'Rahul'],
    'Marks': [85, 90, 78]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Branch': ['CSE', 'ECE', 'IT']
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

# 1. Inner Merge
print("\nInner Merge:")
print(pd.merge(df1, df2, on='ID', how='inner'))

# 2. Left Merge
print("\nLeft Merge:")
print(pd.merge(df1, df2, on='ID', how='left'))

# 3. Right Merge
print("\nRight Merge:")
print(pd.merge(df1, df2, on='ID', how='right'))

# 4. Outer Merge
print("\nOuter Merge:")
print(pd.merge(df1, df2, on='ID', how='outer'))

# 5. Concatenate Rows
print("\nRow Concatenation:")
print(pd.concat([df1, df1], ignore_index=True))

# 6. Concatenate Columns
print("\nColumn Concatenation:")
print(pd.concat([df1, df2], axis=1))

# 7. Join using ID as index
left = df1.set_index('ID')
right = df2.set_index('ID')

print("\nJoin:")
print(left.join(right, how='inner'))
```
