```python

import pandas as pd

# Create DataFrame 1
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ravi', 'Sita', 'Rahul'],
    'Marks': [85, 90, 78]
})

# Create DataFrame 2
df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Branch': ['CSE', 'ECE', 'IT']
})

print("DATAFRAME 1")
print(df1)

print("\nDATAFRAME 2")
print(df2)

# 1. HIERARCHICAL INDEXING
print("\n1. HIERARCHICAL INDEXING")

multi_df = df1.set_index(['ID', 'Name'])
print(multi_df)

# 2. COMBINING USING CONCAT
print("\n2. COMBINING DATAFRAMES")

combined = pd.concat([df1, df1], ignore_index=True)
print(combined)

# 3. MERGING DATAFRAMES
print("\n3. MERGING DATAFRAMES")

merged = pd.merge(df1, df2, on='ID', how='inner')
print(merged)

# 4. RESHAPING USING MELT
print("\n4. RESHAPING USING MELT")

long_df = pd.melt(
    df1,
    id_vars=['ID', 'Name'],
    value_vars=['Marks'],
    var_name='Subject',
    value_name='Score'
)
print(long_df)

# 5. PIVOTING
print("\n5. PIVOTING")

sales = pd.DataFrame({
    'Region': ['East', 'East', 'West', 'West'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Phone'],
    'Sales': [50000, 20000, 45000, 25000]
})

pivot_df = sales.pivot(
    index='Region',
    columns='Product',
    values='Sales'
)
print(pivot_df)

# 6. PIVOT TABLE
print("\n6. PIVOT TABLE")

sales2 = pd.concat([
    sales,
    pd.DataFrame({
        'Region': ['East', 'West'],
        'Product': ['Laptop', 'Phone'],
        'Sales': [10000, 5000]
    })
], ignore_index=True)

pivot_table = sales2.pivot_table(
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='sum'
)
print(pivot_table)

# 7. STACK AND UNSTACK
print("\n7. STACK")

stacked = pivot_df.stack()
print(stacked)

print("\n8. UNSTACK")

unstacked = stacked.unstack()
print(unstacked)

print("\nExperiment completed successfully.")
```
